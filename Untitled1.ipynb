```python
import string
text ="! hey, i am *$# , "

tokens = word_tokenize(text)
new_text = []
for i in tokens:
    if i not in string.punctuation:
        new_text.append(i)

new_text
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

text = "Hello, my name is Himanshi. I am learning NLP."
tokens=word_tokenize(text)
clean_text=[]
for i in tokens:
    if i.lower() not in stopwords.words('english'):
        clean_text.append(i)

clean_text
```

```python
# Stemming (fast)
from nltk.stem import PorterStemmer
stemming=PorterStemmer()

words = ["playing","plays","played","unplayed","player","players","playful"]
for i in words:
    print(i,"->",stemming.stem(i.lower()))
```

```python
# Lemmatization (slow)
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemm = WordNetLemmatizer()
words=["fairly","easily","happy","happiness","unplayed","played"]
for i in words:
    print(lemm.lemmatize(i))
```

```python
# lemmatization in sentence
from nltk.stem import WordNetLemmatizer
Lemm = WordNetLemmatizer()

text = """ It was his swimming finals today, but he was running late for the bus.
so he ran to bus stop , and at last he won the finale"""
kavish=word_tokenize(text)

for i in kavish:
    print( i,"->",Lemm.lemmatize(i, pos='v'))
```
